```python
import pandas as pd

df = pd.DataFrame(items)
print("Jumlah data:", len(df))
print("Jumlah kolom:", len(df.columns))
print("Nama kolom:", df.columns.tolist())
df[['uniqueId', 'text', 'createTimeISO', 'diggCount']].head(1000)
```

**Simpan teks asli**

```python
data = df.copy()
data['text_original'] = data['text']
data[['text_original']].head(1000)
```

**Lowercase**

```python
data['text_clean'] = data['text'].astype(str).str.lower()
data[['text_original', 'text_clean']].head(1000)
```

**Hapus URL**

```python
import re

data['text_clean'] = data['text_clean'].apply(
    lambda x: re.sub(r'http\S+|www\S+|https\S+', '', x)
)
data[['text_original', 'text_clean']].head(1000)
```

**Hapus mention & hashtag**

```python
data['text_clean'] = data['text_clean'].apply(
    lambda x: re.sub(r'@\w+|#\w+', '', x)
)
data[['text_original', 'text_clean']].head(1000)
```

**Hapus tanda baca**

```python
data['text_clean'] = data['text_clean'].apply(
    lambda x: re.sub(r'[^\w\s]', '', x)
)
data[['text_original', 'text_clean']].head(1000)
```

**Hapus angka**

```python
data['text_clean'] = data['text_clean'].apply(
    lambda x: re.sub(r'\d+', '', x)
)
data[['text_original', 'text_clean']].head(1000)
```

**Rapikan spasi**

```python
data['text_clean'] = data['text_clean'].apply(
    lambda x: re.sub(r'\s+', ' ', x).strip()
)
data[['text_original', 'text_clean']].head(1000)
```

**Hapus komentar kosong dan duplikat**

```python
print("Sebelum:", len(data))
data = data[data['text_clean'].str.strip() != ''].copy()
data = data.drop_duplicates(subset='text_clean').reset_index(drop=True)
print("Jumlah komentar setelah cleaning:", len(data))
```

**Tokenisasi**

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

data['tokens'] = data['text_clean'].apply(word_tokenize)
data[['text_clean', 'tokens']].head(1000)
```

**Stopword Indonesia**

```python
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

factory = StopWordRemoverFactory()
stopwords = set(factory.get_stop_words())
print("Jumlah stopword:", len(stopwords))
```

**Hapus stopword**

```python
data['tokens_no_stopword'] = data['tokens'].apply(
    lambda tokens: [w for w in tokens if w not in stopwords]
)
data[['tokens', 'tokens_no_stopword']].head(1000)
```

**Gabung jadi teks**

```python
data['text_no_stopword'] = data['tokens_no_stopword'].apply(lambda t: ' '.join(t))
data[['text_clean', 'text_no_stopword']].head(1000)
```

**Siapkan stemmer**

```python
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

stemmer = StemmerFactory().create_stemmer()
```

**Stemming (versi cepat dengan cache)**

```python
cache = {}

def stem_cepat(teks):
    hasil = []
    for kata in teks.split():
        if kata not in cache:
            cache[kata] = stemmer.stem(kata)
        hasil.append(cache[kata])
    return ' '.join(hasil)

data['text_stemmed'] = data['text_no_stopword'].apply(stem_cepat)
data[['text_no_stopword', 'text_stemmed']].head(1000)
```

**Kolom prepro**

```python
data['prepro'] = data['text_stemmed']
data[['text_original', 'prepro']].head(1000)
```

**Tabel final + simpan CSV**

```python
hasil_tiktok = data[
    ['uniqueId', 'text_original', 'prepro', 'createTimeISO', 'diggCount']
].copy()

hasil_tiktok = hasil_tiktok[hasil_tiktok['prepro'].str.strip() != ''].reset_index(drop=True)

print("Jumlah data final:", len(hasil_tiktok))
hasil_tiktok.to_csv('hasil_tiktok.csv', index=False)
hasil_tiktok.head(1000)
```
