```python
import re
```

**1**. **Ekstraksi ID Transaksi**

```python
text = "Pembayaran untuk pesanan #INV-2026-08912 telah berhasil diverifikasi pada pukul 14.30 WIB."
pattern = r"#(INV-\d{4}-\d+)"

hasil = re.findall(pattern, text)
print("Hasil:", hasil)
```

Penjelasan : cari tanda pagar #, lalu setelah itu harus ada tulisan "INV-", lalu 4 angka (\d{4}), lalu strip, lalu angka lagi sebanyak apapun (\d+). Karena ada tanda kurung () di sekeliling "INV-\d{4}-\d+", Python cuma mengambil apa yang ada di dalam kurung itu, tanda pagarnya sendiri dibuang.

**2**. **Ekstraksi Nominal Harga (Rupiah)**

```python
text = "Total belanjaan Anda adalah Rp 450.000 dengan diskon promo sebesar Rp 50.000."
pattern = r"Rp\s?[\d.]+"

hasil = re.findall(pattern, text)
print("Hasil:", hasil)
```

Penjelasan : cari kata "Rp", boleh diikuti satu spasi atau tidak sama sekali (\s?), lalu setelah itu ambil sebanyak mungkin karakter yang berupa angka atau titik ([\d.]+).

Titik di sini dianggap "karakter yang boleh diambil" tanpa peduli itu titik pemisah ribuan atau titik tanda baca akhir kalimat. Makanya elemen kedua hasilnya jadi 'Rp 50.000.' (ada titik nyantol di ujung) — karena kalimat itu memang diakhiri titik tepat setelah angka terakhir.

**3**. **Masking Nomor Kartu Kredit**

```python
text = "Transaksi menggunakan kartu kredit 4532-1234-5678-9012 telah disetujui."
pattern = r"(\d{4})-(\d{4})-(\d{4})-(\d{4})"

masked = re.sub(pattern, r"XXXX-XXXX-XXXX-\4", text)
print("Hasil:", masked)
```

Penjelasan :  
Nomor kartu dipecah jadi 4 kelompok 4-digit terpisah, masing-masing dibungkus kurung jadi grup 1, 2, 3, 4. Lalu dipakai re.sub dengan teks pengganti XXXX-XXXX-XXXX-\4.

\4 artinya "pakai lagi isi grup ke-4 yang tadi ditangkap". Jadi 3 kelompok pertama diganti paksa jadi "XXXX", tapi kelompok ke-4 (4 digit terakhir kartu) tetap dikembalikan apa adanya. Ini teknik standar untuk masking data sensitif — cukup tampilkan 4 digit terakhir saja.

**4**. **Validasi Kode Promo/Voucher**

```python
text = "Gunakan kode promo HEMAT50K atau DISKON100 untuk mendapatkan potongan harga."
pattern = r"\b[A-Z]+\d+[A-Z]*\b"

hasil = re.findall(pattern, text)
print("Hasil:", hasil)
```

Penjelasan : \b artinya batas kata (jadi tidak nyangkut di tengah kata lain). Isinya: huruf besar satu atau lebih ([A-Z]+), lalu angka satu atau lebih (\d+), lalu opsional huruf besar lagi di belakang ([A-Z]*, boleh nol kali). Pola ini pas untuk kode promo yang biasanya "HURUF+ANGKA" atau "HURUF+ANGKA+HURUF".

**5**. **Identifikasi Metode Pembayaran (E-Wallet/Bank)**

Penjelasan :
Ini pattern dengan dua "kemungkinan besar" dipisah tanda | di level paling luar: kemungkinan pertama cocokkan nama metode bayar (Transfer Bank + nama banknya, atau nama e-wallet), kemungkinan kedua cocokkan angka Virtual Account setelah "VA:".

karena dua kemungkinan itu masing-masing punya kurungnya sendiri (grup 1 dan grup 2), setiap kali satu kemungkinan yang cocok, grup yang lain otomatis kosong. Makanya hasilnya berupa pasangan (tuple) dengan string kosong.

**9**. **Deteksi Berat Produk**

```python
text = "Berat bersih item ini adalah 500gr, sedangkan berat packing sekitar 0.75kg."
pattern = r"\d+(?:\.\d+)?\s?(?:gr|kg)"

hasil = re.findall(pattern, text)
print("Hasil:", hasil)
```

Karena desimalnya opsional, pattern ini cocok baik untuk "500gr" (bulat) maupun "0.75kg" (desimal).

**10**. **Parsing Varian Warna dan Ukuran**

```python
text = "Kaos Polos Cotton 30s - Warna: Merah Cabai - Size: XL."
pattern = r"Warna:\s*([\w\s]+?)\s*-\s*Size:\s*(\w+)"

m = re.search(pattern, text)
print("Warna :", m.group(1))
print("Size  :", m.group(2))
```

Penjelasan :
Dipakai re.search (cari satu match pertama saja, beda dari re.findall yang cari semua) karena hasilnya mau diambil per-grup pakai .group(1) dan .group(2).

Grup 1: teks setelah "Warna:" sampai ketemu " - Size:" — pakai tanda +? (lazy) supaya berhenti secepat mungkin, tidak "kebablasan" mengambil terlalu banyak. Grup 2: kata setelah "Size:".

**13**. **Sanitasi Nomor Telepon Kurir/Pelanggan**

```python
text = "Hubungi kurir di 0812-3456-7890 atau +6281298765432 jika kendala."
pattern = r"(?:\+62|0)\d{2,4}(?:-?\d{3,4}){2,3}"

ditemukan = re.findall(pattern, text)
disensor = re.sub(pattern, "[NOMOR_DISENSOR]", text)
print("Ditemukan:", ditemukan)
print("Disensor :", disensor)
```

Penjelasan :
Nomor telepon Indonesia bisa diawali "0" atau "+62" — dua kemungkinan ini ditulis dalam (?:\+62|0). Setelah itu, 2-4 digit, lalu diikuti pola berulang "strip opsional + 3-4 digit" sebanyak 2 sampai 3 kali. Pattern ini cukup fleksibel menangani dua gaya penulisan sekaligus: yang pakai strip ("0812-3456-7890") dan yang tidak ("+6281298765432").

Setelah ditemukan, re.sub mengganti semua nomor yang ketemu jadi tulisan "[NOMOR_DISENSOR]" — teknik umum untuk menyembunyikan data pribadi di log atau tampilan publik.

**18**. **Filter Kata Kasar/Spam di Ulasan**

```python
text = "Barang jelek banget, penipu ini seller ajg penipuan."
bad_words = ["ajg", "anjing", "kontol", "goblok", "penipu", "tolol", "bangsat"]
pattern = r"\b(" + "|".join(bad_words) + r")\w*\b"

ditemukan = re.findall(pattern, text, flags=re.IGNORECASE)
disensor = re.sub(pattern, lambda m: "*" * len(m.group()), text, flags=re.IGNORECASE)
print("Ditemukan:", ditemukan)
print("Disensor :", disensor)
```

Penjelasan :
Pattern-nya tidak ditulis manual, tapi dirakit pakai kode: gabungkan semua kata di bad_words dengan tanda | di antaranya, lalu tambahkan \w* di belakang supaya imbuhan ikut tertangkap juga (misalnya "penipu" jadi tetap kena walau di teks ditulis "penipuan").

Untuk menyensor, dipakai re.sub dengan fungsi, bukan teks tetap:lambda m: "*" * len(m.group())
  yang dimana, setiap kata yang ketemu diganti bintang sebanyak panjang katanya sendiri — jadi jumlah bintangnya menyesuaikan, bukan jumlah tetap.

flags=re.IGNORECASE membuat pencarian tidak peduli huruf besar/kecil.

**29**. **Deteksi Percobaan SQL Injection di Kolom Pencarian**

```python
text = "User mencari keyword: baju pria' OR '1'='1"
pattern = r"(?i)('\s*(?:or|and)\s*'?\w+'?\s*=\s*'?\w+'?|--|;\s*--|\bunion\b|\bselect\b|\bdrop\b)"

terdeteksi = bool(re.search(pattern, text))
match = re.findall(pattern, text)
print("Terdeteksi:", terdeteksi)
print("Match     :", match)
```

Penjelasan :
(?i) di awal artinya "abaikan besar/kecil huruf untuk seluruh pattern ini" (sama efeknya dengan flags=re.IGNORECASE, tapi ditulis langsung di dalam pattern). Setelah itu ada beberapa kemungkinan pola dipisah |, masing-masing adalah ciri khas serangan SQL injection:

kutip diikuti "OR"/"AND" lalu perbandingan nilai ('...OR...=...') — pola klasik "bypass login"
-- atau ;-- — tanda komentar SQL, dipakai untuk "mematikan" sisa query
kata kunci berbahaya seperti union, select, drop

```python
text = "Barangnya bagus banget kak! \U0001F44D\U0001F525 Recommended seller! \u2764\uFE0F"
pattern = r"[\U0001F300-\U0001FAFF\U00002600-\U000027BF\U0001F1E6-\U0001F1FF]"

hasil = re.findall(pattern, text)
print("Hasil:", hasil)
```

Penjelasan :
Ini bukan mendeteksi emoji berdasarkan "tampilannya", tapi berdasarkan rentang kode Unicode resmi tempat emoji didefinisikan. Ada 3 rentang digabung dalam satu character class [...]:

1. \U0001F300 sampai \U0001FAFF — blok simbol & piktograf (mencakup 👍 dan 🔥)
2. \U00002600 sampai \U000027BF — blok simbol lain-lain & dingbat
3. \U0001F1E6 sampai \U0001F1FF — blok huruf regional (dipakai untuk bendera negara)

**39**. **Parsing Nama dan Nilai Atribut Spesifikasi (JSON-like text)**

```python
text = "Spesifikasi: [RAM: 8GB], [Storage: 256GB], [Baterai: 5000mAh]"
pattern = r"\[(\w+):\s*([\w\d]+)\]"

hasil = re.findall(pattern, text)
print("Hasil:", hasil)
```

Penjelasan :
Mirip konsep dengan nomor 34, tapi kali ini ada dua grup: grup 1 nama atributnya (sebelum titik dua), grup 2 nilainya (setelah titik dua). Karena tiga blok "[Nama: Nilai]" ada di teks, findall mengembalikan tiga tuple sekaligus.

**40**.**(Bonus) Normalisasi Kata Baku Singkatan Chat**

```python
text = "brg sdh smp dgn slmt, tq seller"
kamus = {
    "brg": "barang", "sdh": "sudah", "smp": "sampai", "dgn": "dengan",
    "slmt": "selamat", "tq": "terima kasih", "yg": "yang", "gk": "tidak",
    "ga": "tidak", "aja": "saja", "bgt": "banget", "utk": "untuk",
    "dr": "dari", "krn": "karena", "jd": "jadi", "dpt": "dapat",
}
pattern = r"\b\w+\b"

def ganti(m):
    kata = m.group(0)
    return kamus.get(kata.lower(), kata)

hasil = re.sub(pattern, ganti, text)
print("Sebelum:", text)
print("Sesudah:", hasil)
```

Penjelasan :
1. Untuk setiap kata yang match pattern (yaitu semua kata dalam kalimat, karena \b\w+\b sangat umum), Python memanggil fungsi ganti.
2. Di dalam fungsi, m.group(0) mengambil kata asli yang sedang diproses.
3. kamus.get(kata.lower(), kata) mengecek: kalau kata itu (versi huruf kecil) ada di dalam kamus singkatan, kembalikan bentuk bakunya. Kalau tidak ada di kamus, kembalikan kata aslinya tanpa perubahan (ini fungsi parameter kedua pada .get() — nilai default kalau key tidak ditemukan).
4. Hasil dari fungsi ini otomatis dipakai re.sub sebagai pengganti tiap kata.
